```python
import numpy as np 

def baud_rate(fclk_Mhz, divisor):
    # 1e6 correctly converts MHz to Hz
    return (fclk_Mhz * 1e6) / (4 * divisor) 

fclks = np.array([10, 20, 25, 30, 40, 50, 60, 70, 75, 80, 90, 100])
divs = np.array([1302, 217, 108, 54, 27, 22, 19, 16, 15, 10, 8, 6, 5, 4, 2])

# Print header row for frequencies
header = "Divisor | " + " | ".join(f"{fclk:6}MHz" for fclk in fclks)
print(header)
print("-" * len(header))

# Print rows for each divisor
for div in divs:
    row_rates = []
    for fclk in fclks:
        # Convert Hz to kHz
        rate_khz = baud_rate(fclk, div) / 1e3
        row_rates.append(f"{rate_khz:8.2f}")
    
    print(f"{div:<7} | " + " | ".join(row_rates))
```
